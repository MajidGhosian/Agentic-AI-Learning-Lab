# 03 — ReAct Agent Demo

This notebook demonstrates the `ReActAgent` from `react_agent.py`.

It runs a multi-step question that requires **two tool calls** (a search
lookup and a calculation) chained together via the Thought -> Action ->
Observation loop.

> Requires `ANTHROPIC_API_KEY` to be set in your environment before running.

```python
import os
import sys

sys.path.append(os.getcwd())

from react_agent import ReActAgent

assert os.environ.get("ANTHROPIC_API_KEY"), "Set ANTHROPIC_API_KEY first"

agent = ReActAgent(verbose=True)
```

## Example 1 — pure calculation
Only needs the `calculator` tool.

```python
answer = agent.run("What is 12% of 850, plus the number of letters in the word 'agentic'?")
print("\nFINAL:", answer)
```

## Example 2 — search + calculator chained
Requires looking up a fact, then reasoning over it.

```python
answer = agent.run(
    "Look up what the ReAct pattern is, then tell me how many words are in its short description."
)
print("\nFINAL:", answer)
```

## What to look for in the printed trace

- Each `Thought:` line shows the model deciding what to do next.
- Each `Action: tool[input]` line shows a real Python function being called
  (see `tools.py`) — not a hallucinated result.
- Each `Observation:` line is the *actual* return value from that Python
  function, fed back into the model's next turn.
- The loop repeats until the model is confident enough to emit
  `Final Answer:`.
